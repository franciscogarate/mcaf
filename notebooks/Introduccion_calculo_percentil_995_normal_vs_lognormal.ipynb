### Cálculo percentil 99.5% distribución normal
$\mu = 0$ y $\sigma=1$

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

```python
#distr_normal = st.norm(loc=4,scale=1)
distr_normal = st.norm()
```

```python
x = np.linspace(-4,4,100)
y = distr_normal.pdf(x)
plt.plot(x,y)
```

## Caso distribución lognormal

```python
distr_normal.ppf(0.995)
```

Ese “3” sale de una lognormal calibrada/normalizada (normalmente con media 1), no de la lognormal base LogNormal(0,1).

```python
distr_lognormal = st.lognorm(s=0.45, scale=1)
distr_lognormal.ppf(0.995)
```
